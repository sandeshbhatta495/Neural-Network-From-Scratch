```python
# 3 inputs coming into this neuron
inputs = [1.0, 2.0, 3.0]

# 3 weights — one per input
weights = [0.2, 0.8, -0.5]

# one bias — one per neuron
bias = 2.0

# manual computation
output = (inputs[0]*weights[0] +
          inputs[1]*weights[1] +
          inputs[2]*weights[2] + bias)

print(output)  # 2.3
```

```python
# the Universal rule 
# inputs:      (batch_size, n_features)     = (3, 4)
# weights:     (n_neurons, n_features)      = (3, 4)
# weights.T:   (n_features, n_neurons)      = (4, 3)
# dot result:  (batch_size, n_neurons)      = (3, 3)
# biases:      (1, n_neurons) or (n_neurons,) = (3,)
# output:      (batch_size, n_neurons)      = (3, 3)
```

```python
inputs = np.random.randn(5, 3)   # 5 samples, 3 features
weights = np.random.randn(4, 3)  # 4 neurons
output = np.dot(inputs, weights.T)
```
